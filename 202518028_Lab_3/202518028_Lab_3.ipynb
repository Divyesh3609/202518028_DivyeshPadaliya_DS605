```python
outlier_cols = [
    "lead_time",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies",
    "previous_cancellations",
    "adr"
]

before = len(hotel)

for col in outlier_cols:

    Q1 = hotel[col].quantile(0.25)
    Q3 = hotel[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    hotel = hotel[
        (hotel[col] >= lower) &
        (hotel[col] <= upper)
    ]

after = len(hotel)

print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
X = hotel.drop(columns=['is_canceled'], axis=1)
y = hotel["is_canceled"]

num_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
cat_cols = X.select_dtypes(include=["object", "category"]).columns.tolist()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

num_pipeline_A = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

num_pipeline_B = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])
```

```python
preprocessor_A = ColumnTransformer([
    ("num", num_pipeline_A, num_cols),
    ("cat", cat_pipeline, cat_cols)
])

preprocessor_B = ColumnTransformer([
    ("num", num_pipeline_B, num_cols),
    ("cat", cat_pipeline, cat_cols)
])
```

```python
pipeline_A = Pipeline([
    ("preprocessor", preprocessor_A)
])

pipeline_B = Pipeline([
    ("preprocessor", preprocessor_B)
])
```

```python
X_train_A = pipeline_A.fit_transform(X_train)
X_test_A = pipeline_A.transform(X_test)

X_train_B = pipeline_B.fit_transform(X_train)
X_test_B = pipeline_B.transform(X_test)
```

# **Part B.**

## Task 4.

```python
from sklearn.linear_model import LogisticRegression

model_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("classifier", LogisticRegression(max_iter=1000))
])

model_A.fit(X_train, y_train)
```

```python
model_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("classifier", LogisticRegression(max_iter=1000))
])

model_B.fit(X_train, y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier

dt_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("classifier", DecisionTreeClassifier(random_state=42))
])

dt_A.fit(X_train, y_train)

dt_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("classifier", DecisionTreeClassifier(random_state=42))
])

dt_B.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def evaluate_model(model, X_train, y_train, X_test, y_test):

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    train_accuracy = accuracy_score(y_train, train_pred)

    test_accuracy = accuracy_score(y_test, test_pred)
    precision = precision_score(y_test, test_pred)
    recall = recall_score(y_test, test_pred)
    f1 = f1_score(y_test, test_pred)

    print("Training Accuracy:", train_accuracy)
    print("Testing Accuracy :", test_accuracy)
    print("Precision         :", precision)
    print("Recall            :", recall)
    print("F1-Score          :", f1)
```

```python
print("Logistic Regression - Pipeline A")
evaluate_model(model_A, X_train, y_train, X_test, y_test)
```

```python
print("Logistic Regression - Pipeline B")
evaluate_model(model_B, X_train, y_train, X_test, y_test)
```

```python
print("Decision Tree - Pipeline A")
evaluate_model(dt_A, X_train, y_train, X_test, y_test)
```

```python
print("Decision Tree - Pipeline B")
evaluate_model(dt_B, X_train, y_train, X_test, y_test)
```

```python
def get_metrics(model, X_train, y_train, X_test, y_test):
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    return {
        "Training Accuracy": accuracy_score(y_train, train_pred),
        "Testing Accuracy": accuracy_score(y_test, test_pred),
        "Precision": precision_score(y_test, test_pred),
        "Recall": recall_score(y_test, test_pred),
        "F1-Score": f1_score(y_test, test_pred)
    }

results = []

experiments = [
    ("Logistic Regression", "Pipeline A", model_A),
    ("Logistic Regression", "Pipeline B", model_B),
    ("Decision Tree", "Pipeline A", dt_A),
    ("Decision Tree", "Pipeline B", dt_B)
]

for model_name, pipeline_name, model in experiments:
    metrics = get_metrics(
        model,
        X_train, y_train,
        X_test, y_test
    )

    results.append({
        "Model": model_name,
        "Pipeline": pipeline_name,
        **metrics
    })

comparison_df = pd.DataFrame(results)

comparison_df
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

lr_results = comparison_df[
    comparison_df["Model"] == "Logistic Regression"
]

best_lr = lr_results.loc[
    lr_results["Testing Accuracy"].idxmax()
]

dt_results = comparison_df[
    comparison_df["Model"] == "Decision Tree"
]

best_dt = dt_results.loc[
    dt_results["Testing Accuracy"].idxmax()
]

print("Best Logistic Regression:")
print(best_lr)

print("\nBest Decision Tree:")
print(best_dt)
```
