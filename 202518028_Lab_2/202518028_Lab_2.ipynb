```python
pass_count_survival = train.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("Survived", "count"),
    Survival_Rate=("Survived", "mean")
)

pass_count_survival["Survival_Rate"] = pass_count_survival["Survival_Rate"] * 100

print(pass_count_survival['Passenger_Count'])
print()
print(pass_count_survival['Survival_Rate'])
```

```python
group_embark = train.groupby("Embarked").agg(
    Passenger_Count=("Survived", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

group_embark["Survival_Rate"] *= 100

print(group_embark['Passenger_Count'])
print()
print(group_embark['Average_Fare'])
print()
print(group_embark['Survival_Rate'])
```

## *Task 7.*

```python
missing = train.isnull().sum()
missing
```

```python
missing_per = train.isnull().mean() * 100
missing_per
```

```python
miss_age = train['Age'].isnull().sum()
print('Before filling missing values: ', miss_age)
```

```python
mean_age = train['Age'].mean()
train['Age'] = train['Age'].fillna(mean_age)
fill_age = train['Age'].isnull().sum()
print('After filling missing values: ', fill_age)
```

```python
NaN_values = train.isna().sum()
NaN_values
```

```python
train['Cabin'] = train['Cabin'].fillna('Unknown')
train['Cabin'].isna().sum()
```

```python
train['Embarked'] = train['Embarked'].fillna(train['Embarked'].mode()[0])
train['Embarked'].isna().sum()
```

```python
train.isnull().sum()
```

```python
Q1 = train["Fare"].quantile(0.25)
Q3 = train["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = train[
    (train["Fare"] < lower_bound) |
    (train["Fare"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of outliers:", len(outliers))
```

```python
train['FamilySize'] = train['SibSp'] + train['Parch'] + 1
train['IsAlone'] = (train['FamilySize']==1).astype(int)
train.head()
```

```python
pivot = pd.pivot_table(
    train,
    values='Survived',
    index='Sex',
    columns='Pclass',
    aggfunc='mean'
)
print(pivot)
```

```python
highest = pivot.stack().idxmax()
highest_value = pivot.stack().max()

lowest = pivot.stack().idxmin()
lowest_value = pivot.stack().min()

print("Highest:", highest, highest_value)
print("Lowest:", lowest, lowest_value)
```

## *Task 9.*

```python
num_cols = ['Survived','Pclass','Age','SibSp','Parch','Fare']
corr = train[num_cols].corr()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', center=0)
plt.title('Correlation HeatMap')
plt.show()
```

```python
corr_pairs = corr.where(
    ~np.eye(corr.shape[0], dtype=bool)
).stack()

strongest_positive = corr_pairs.idxmax()
positive_value = corr_pairs.max()

strongest_negative = corr_pairs.idxmin()
negative_value = corr_pairs.min()

print("Strongest positive:", strongest_positive, "=", round(positive_value, 2))
print("Strongest negative:", strongest_negative, "=", round(negative_value, 2))
```

```python
survival_rate = train.groupby('Sex')['Survived'].mean()

plt.figure(figsize=(6, 4))
sns.barplot(x=survival_rate.index, y=survival_rate.values)

plt.title('Survival Rate by Sex')
plt.xlabel('Sex')
plt.ylabel('Survival Rate')
plt.ylim(0, 1)

for i, value in enumerate(survival_rate.values):
    plt.text(i, value + 0.02, f'{value * 100:.1f}%', ha='center')

plt.show()

print("Female survival rate:", f"{survival_rate['female'] * 100:.1f}%")
print("Male survival rate:", f"{survival_rate['male'] * 100:.1f}%")
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=train, x='Age', y='Fare', hue='Survived')
plt.title('Age vs. Fare with Survival')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

### Observation:
- Female passengers had higher survival rate than male passengers.
- Female passengers survival rate was 74.2% and Male passengers survival rate was 18.9%
- Passengers traveled in 1st class has higher survival rate than others.
- Pclass and Fare has strongest negative correlation(-0.55)
- SibSp and Parch has strongest positive correlation(0.41)
- Most passengers had traveled with lower fares.
